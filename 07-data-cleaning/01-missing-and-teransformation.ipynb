## Missing Data

```python
import pandas as pd
import numpy as np
```

```python
# Creating a pandas Series with some missing data
string_data = pd.Series(['aardvark', 'artichoke', np.nan, 'avocado'])

# Assign None to first element
string_data[0] = None

# Check which values are missing
print(string_data.isnull())
print(string_data.isnull().sum)
```

```python
# Import NaN as NA for shorter code
from numpy import nan as NA
data = pd.Series([1, NA, 3.5, NA, 7])

# Drop all missing values
print(data.dropna())

# Alternative way using boolean indexing
data[data.notnull()]
```

```python
data = pd.DataFrame([[1., 6.5, 3.], [1., NA, NA],
                     [NA, NA, NA], [NA, 6.5, 3.]])

# Only drop rows where ALL values are NA
data.dropna(how='all')
```

```python
# Create a DataFrame with random numbers
df = pd.DataFrame(np.random.randn(7, 3))

# In[28]: Set some values to NA
df.iloc[:4, 1] = NA 
df.iloc[:2, 2] = NA 

# View the DataFrame
df
```

```python
# Fill all NaN with 5
df.fillna(5)
```

```python
# Filll each column with different values
df.fillna({1: 0.5, 2: 0})
```

```python
# Forward fill - propagate last valid value forward
df.ffill()
```

```python
# Example of backward fill - propagate next valid value backward
df.bfill()
```

```python
# Fill missing values with the mean of the Series
data.fillna(data.mean())
```

```python
# Example: Fill with median instead of mean
data.fillna(data.median())
```

```python
# Keep rows with at least 2 non-NA values
df.dropna(thresh=2)
```

```python
# Advanced Example: Different strategies for different columns
df_advanced = pd.DataFrame({
    'A': [1, np.nan, 3, np.nan, 5],
    'B': [10, 20, np.nan, 40, np.nan],
    'C': ['x', 'y', np.nan, 'z', np.nan]
})

# Fill numeric columns with mean, categorical with mode
fill_values = {
    'A': df_advanced['A'].mean(),     
    'B': df_advanced['B'].median(),  
    'C': df_advanced['C'].mode()[0] 
}

df_advanced_filled = df_advanced.fillna(fill_values)
df_advanced_filled
```

### **Dropping Missing Values - dropna()**
|Parameter|	Description	|Default|
|---------|-------------|-------|
`axis`	|Drop rows (0) or columns (1)	| `0`
`how`	|`'any'` (drop if any NaN) or 'all' (drop if all NaN)	|`'any'`
`thresh`|	Require at least N non-NA values to keep	| `None`
`subset`|	Columns to consider for NaN detection|	`None`
`inplace`|	Modify the original DataFrame	| `False

### **Filling Missing Values - fillna()**
| Parameter|	Description|	Default|
|----------|---------------|-----------|
`value`|	Scalar value or dict to fill NaN's	|Required
`method`|	`'ffill'` (forward fill) or` 'bfill'` (backward fill)	|`None`
`axis`	|Axis to fill on|	`0`
`inplace`	|Modify without copy	|`False`
`limit`	|Max consecutive periods to fill	|`None`

### **Quick Reference Card**
|Task	|Code|
|-------|----|
Remove all rows with any NaN	|`df.dropna()`
Remove columns with any NaN	|`df.dropna(axis=1)`
Fill all NaN with a fixed value	|`df.fillna(0)`
Forward fill	|`df.ffill()` or `df.fillna(method='ffill')`
Backward fill|`	df.bfill()` or `df.fillna(method='bfill')`
Fill with column mean	|`df.fillna(df.mean())`
Check if any NaN exists	|`df.isnull().any().any()`
Count NaN per column	|`df.isnull().sum()`
Keep rows with at least N non-NA	|`df.dropna(thresh=N)`

    Note: Methods with inplace=True modify the original DataFrame. Use with caution or create a copy first with df.copy().

## Data Transformation

```python
# Create a DataFrame with duplicate rows
data = pd.DataFrame({'k1': ['one', 'two'] * 3 + ['two'],
                     'k2': [1, 1, 2, 3, 3, 4, 4]})

# Check for duplicate rows (returns boolean Series)
print(data.duplicated())

# Remove duplicate rows, keep first occurrence
print(data.drop_duplicates())
```

```python
# Add a new column for demonstration
data['v1'] = range(7)
print(data)

# Remove duplicates based only on 'k1' column
data.drop_duplicates(['k1'])
```

```python
# Keep last occurrence of duplicates
data.drop_duplicates(['k1', 'k2'], keep='last')
```

```python
# Create a DataFrame with meat types
data = pd.DataFrame({
    'food': ['bacon', 'pulled pork', 'bacon', 'Pastrami', 'corned beef', 
             'Bacon', 'pastrami', 'honey ham', 'nova lox'],
    'ounces': [4, 3, 12, 6, 7.5, 8, 3, 5, 6]
})

# Define a mapping from meat to animal
meat_to_animal = {
    'bacon': 'pig',
    'pulled pork': 'pig',
    'pastrami': 'cow',
    'corned beef': 'cow',
    'honey ham': 'pig',
    'nova lox': 'salmon'
}

# Convert all food names to lowercase
lowercased = data['food'].str.lower()

# Map food names to animal names using the dictionary
data['animal'] = lowercased.map(meat_to_animal)
data
```

```python
# Use lambda function to combine lowercase conversion and mapping in one step
data['food'].map(lambda x: meat_to_animal[x.lower()])
```


#### **Working with Duplicate Records**
`duplicated()` - Detect duplicate rows
- Returns Boolean Series (True for duplicates)
- By default, marks all duplicates except the first occurrence

|Parameter|	Description|	Default|
|---------|------------|-----------|
`subset`|	Columns to consider for duplicate detection	|All columns
`keep`	|Which occurrence to keep: `'first'`, `'last'`, or False (remove all)	| '`first`'
`inplace`	|Modify the original DataFrame	| Fa`lse
`ignore_index`	|Reset index after dropping	| `False`

#### **Comparison: `map()` vs `apply()` vs `replace()`**
|Method	|Use Case	|Example|
|-------|-----------|-------|
`Series.map()`	|Element-wise mapping (Series only)	|`s.map({'A': 1})`
`Series.apply()`	|Element-wise with function|	`s.apply(lambda x: func(x))`
`DataFrame.apply()`	|Apply function to rows/columns|`	df.apply(func, axis=1)`
`DataFrame.replace()`	|Replace values in entire DataFrame|	`df.replace({'old': 'new'})`

#### Pro Tips:

    Use duplicated() first to inspect before dropping
    Always check the impact of keep=False (removes all duplicates)
    Use str.lower() before mapping to ensure case-insensitive matching
    Lambda functions are great for quick transformations but use named functions for complex logic


### Replacing Values

```python
# Create a Series with sentinel values (markers for missing data)
data = pd.Series([1., -999., 2., -999., -1000., 3.])

# Create a new Series unless use inplace=True
data.replace([-999, -1000], np.nan)
```

```python
# Replace each value with a different replacement using lists
data.replace([-999, -1000], [np.nan, 0])
```

```python
# Replace values using a dictionary (key = old value, value = new value)
data.replace({-999: np.nan, -1000: 0})
```

```python
# Create a DataFrame with custom index and columns
data = pd.DataFrame(np.arange(12).reshape((3, 4)),
                    index=['Ohio', 'Colorado', 'New York'],
                    columns=['one', 'two', 'three', 'four'])
print(data)

# Define lambda to take first 4 characters and uppercase them
transform = lambda x: x[:4].upper()

# Apply transformation to index using map
data.index.map(transform)
```

```python
# Assign transformed index to the DataFrame (in-place modification)
data_index= data.index = data.index.map(transform)
print(data_index)
```

```python
# Use rename to create a new DataFrame with transformed labels
print(data.rename(index=str.title, columns=str.upper))
```

```python
# Use rename with dictionary to change specific index and column labels
data_rename = data.rename(index={'OHIO': 'INDIANA'},
            columns={'three': 'peekaboo'})
print(data_rename)

# Modify the DataFrame in-place (no new object created)
data.rename(index={'OHIO': 'INDIANA'}, inplace=True)
print(data)
```

### Discretization and Binning

```python
# Age data of individuals
ages = [20, 22, 25, 27, 21, 23, 37, 31, 61, 45, 41, 32]

# Define bin edges (boundaries of each group)
bins = [18, 25, 35, 60, 100]

# Discretize ages into these bins
pd.cut(ages, bins) # right=Falce : [18, 25]
```

```python
group_names = ['Youth', 'YoungAdult', 'MiddleAged', 'Senior']
pd.cut(ages, bins, labels=group_names)
```

```python
# Split into 4 equal-width bins (2 decimal precision)
data = np.random.rand(20)
pd.cut(data, 4, precision=2)
```

```python
data = np.random.randn(1000)

# Split into 4 quartiles
cats = pd.qcut(data, 4)

# Value counts (each should have ~250)
pd.Series(cats).value_counts()
```

```python
# Custom percentiles: 0th, 10th, 50th, 90th, 100th
pd.qcut(data, [0, 0.1, 0.5, 0.9, 1.])
```

```python
# English: Read movies file with '::' separator
mnames = ['movie_id', 'title', 'genres']
movies = pd.read_csv('data/movie.csv', header=None, names=mnames)
movies
```

```python
# Extract list to collect all genres
all_genres = []
for x in movies.genres:
    all_genres.extend(x.split('|'))

genres = np.unique(all_genres)
genres
```

```python
# Zero matrix with rows = number of movies, columns = number of genres
zero_matrix = np.zeros((len(movies), len(genres)))

# Convert to DataFrame with column names = genre names
dummies = pd.DataFrame(zero_matrix, columns=genres)
dummies
```

```python
# Iterate over all movies with row index
for i, gen in enumerate(movies.genres):
    # Find column positions of each genre in genres list
    indices = dummies.columns.get_indexer(gen.split('|'))
    
    # Set 1 in those columns for row i
    dummies.iloc[i, indices] = 1
dummies
```

| Method | Description | Syntax Example | Example Output |
|--------|-------------|----------------|----------------|
| `count(sub)` | Return number of non-overlapping occurrences of sub | `"hello hello".count("hello")` | `2` |
| `endswith(suffix)` | Return `True` if string ends with suffix | `"hello.py".endswith(".py")` | `True` |
| `startswith(prefix)` | Return `True` if string starts with prefix | `"hello.py".startswith("hell")` | `True` |

| `rfind(sub)` | Return position of last occurrence of sub; return `-1` if not found | `"hello hello".rfind("hello")` | `6` |
| `replace(old, new)` | Replace occurrences of `old` with `new` | `"hello".replace("l", "x")` | `"hexxo"` |

| `rstrip()` | Trim whitespace from right end | `"hi  ".rstrip()` | `"hi"` |
| `lstrip()` | Trim whitespace from left end | `"  hi".lstrip()` | `"hi"` |
| `split(delim)` | Break string into list of substrings using `delim` | `"a,b,c".split(",")` | `["a", "b", "c"]` |
| `lower()` | Convert all alphabet characters to lowercase | `"HeLLo".lower()` | `"hello"` |
| `upper()` | Convert all alphabet characters to uppercase | `"HeLLo".upper()` | `"HELLO"` |

| `rjust(width)` | Right justify, pad left with spaces to reach `width` | `"hi".rjust(4)` | `"  hi"` |

### Regex 

```python
import re
```

```python
# Text with spaces, tabs, and multiple whitespace characters
text = "foo bar\t baz \tqux"

# Pattern: one or more whitespace characters (space, tab, newline)
re.split(r'\s+', text)
```

```python
# Compile pattern for reuse
regex = re.compile(r'\s+')

# Now we can use it multiple times
regex.split(text) 
```

```python
regex.findall(text)  
```

```python
# Correct way - use r before the string
re.compile(r'\s+') 
```

```python
# Incorrect way (needs double backslash)
re.compile('\\s+') 
```

```python
# A multiline text with emails
text = """Dave dave@google.com
Steve steve@gmail.com
Rob rob@gmail.com
Ryan ryan@yahoo.com
"""

# Email pattern (simplified)
pattern = r'[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}'
regex = re.compile(pattern, flags=re.IGNORECASE)

# Find all emails (list of strings)
regex.findall(text)
```

```python
# Search for the first email (returns a match object)
regex.search(text)
```

```python
# match only works if pattern starts at the very beginning
print(regex.match(text))
```

```python
# Grouping: (username)@(domain).(suffix)
pattern = r'([A-Z0-9._%+-]+)@([A-Z0-9.-]+)\.([A-Z]{2,4})'
regex = re.compile(pattern, flags=re.IGNORECASE)

email = regex.match('wesm@bright.net')
email.groups()   # ('wesm', 'bright', 'net')
```

```python
# See start and end position in original text
email.start()
```

```python
email.end()  
```

```python
text[email.start():email.end()]
```

```python
# findall with groups returns a list of tuples
regex.findall(text)
```

```python
print(regex.sub(r'Username: \1, Domain: \2, Suffix: \3', text))
```
