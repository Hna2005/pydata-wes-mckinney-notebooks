### Figures and Subplots

````python
# Creates empty figure
fig = plt.figure()
````

### Create subplot

````python
# From the book example
fig = plt.figure()

# Add subplots: rows, cols, index
ax1 = fig.add_subplot(2, 2, 1)  # 2 rows, 2 cols, position 1
ax2 = fig.add_subplot(2, 2, 2)  # position 2
ax3 = fig.add_subplot(2, 2, 3)  # position 3

# Plot on each subplot
ax1.plot(np.random.randn(50).cumsum(), 'k--')  # dashed line
ax2.hist(np.random.randn(100), bins=20, color='g', alpha=0.7)  # histogram
ax3.scatter(np.random.randn(100), np.random.randn(100), marker='o', color='r')  # scatter

plt.show()
````

### Simpler method using `subplots()`

````python
# Create figure with subplots in one line
fig, axes = plt.subplots(2, 3) # 2 rows, 3 columns

# axes is a 2x3 array of Axes objects
print(axes.shape)

# Plot on specific subplot
axes[0, 0].plot(np.random.randn(50).cumsum())
axes[1, 2].hist(np.random.randn(100), bins=20)

# Adjust layout
plt.tight_layout()
plt.show()
````

### Adjusting size and spacing

````python
# Control figure size
fig = plt.Figure(figsize=(10, 6))

#Control spacing between subplots
fig, axes = plt.subplots(2, 2, figsize=(8, 8))
plt.subplots_adjust(wspace=0.3, hspace=0.3)
````

````python
# Or with tight_layout (auto adjusts)
fig = plt.figure(figsize=(10, 6))
plt.tight_layout()

# Work with the axes
for i in range(2):
    for j in range(2):
        axes[i, j].plot(np.random.randn(20).cumsum())
        axes[i, j].set_title(f'Subplot {i+1},{j+1}')
````

### Colors, Markes and Line Style

````python
# Line style 
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

# Different line styles
plt.figure(figsize=(10, 6))

# Color and line style shorthand
plt.plot(x, y1, 'g--')      # green dashed line
plt.plot(x, y2, 'b-')       # blue solid line
plt.plot(x, y3, 'r:')       # red dotted line

# With markers
plt.plot(x, y1, 'go-')      # green circles with solid line
plt.plot(x, y2, 'bs--')     # blue squares with dashed line
plt.plot(x, y3, 'r^:')      # red triangles with dotted line

plt.show()
````

````python
# Different ways to specify color
plt.plot(x, y1, color='red')
plt.plot(x, y1, color='#FF5733')      # Hex code
plt.plot(x, y1, color=(0.2, 0.4, 0.6))  # RGB tuple (0-1)
plt.plot(x, y1, color='orange')       # Named color
````

````python
# Example

# Detailed line styling
plt.figure(figsize=(10, 6))

plt.plot(x, y1, 
         color='blue', 
         linestyle='--', 
         linewidth=2,
         marker='o',
         markersize=6,
         markerfacecolor='red',
         markeredgecolor='black',
         markeredgewidth=1,
         label='sin(x)')

plt.plot(x, y2, 
         color='green', 
         linestyle='-', 
         linewidth=3,
         marker='s',
         markersize=5,
         markerfacecolor='yellow',
         label='cos(x)')

plt.legend()
plt.grid(True, alpha=0.3)
plt.title('Advanced Line Plot')
plt.show()
````

## Ticks, Labels and Legends

````python
# Labels

x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, 'b-', linewidth=2)

# Set labels and title
ax.set_title('Sine Wave', fontsize=16, fontweight='bold')
ax.set_xlabel('X axis (radians)', fontsize=12)
ax.set_ylabel('Y axis (amplitude)', fontsize=12)

# Or shorthand
# plt.title('Sine Wave')
# plt.xlabel('X axis')
# plt.ylabel('Y axis')

plt.show()
````

### Ticks

````python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, 'b-', linewidth=2)

# Set tick locations
ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax.set_yticks([-1, -0.5, 0, 0.5, 1])

# Set tick labels
ax.set_xticklabels(['0', r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$'])
ax.set_yticklabels(['-1', '-0.5', '0', '0.5', '1'])

# Rotate x-axis labels for better readability if overlapping
plt.xticks(rotation=45)

plt.show()
````

### Setting the range of axes

````python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, 'b-', linewidth=2)

# Set axis limits
ax.set_xlim(0, 2*np.pi)    # x from 0 to 2π
ax.set_ylim(-1.2, 1.2)     # y from -1.2 to 1.2

# Or shorthand
# plt.xlim(0, 2*np.pi)
# plt.ylim(-1.2, 1.2)

plt.show()
````

### Legend

````python
# From the book example
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y1, 'b-', linewidth=2, label='sin(x)')
ax.plot(x, y2, 'r--', linewidth=2, label='cos(x)')
ax.plot(x, y3, 'g-', linewidth=2, label='sin(x)*cos(x)')

# Add legend
ax.legend()
# Or with custom position
# ax.legend(loc='upper right')  # default
# ax.legend(loc='lower left')
# ax.legend(loc='center')
# ax.legend(loc='best')  # auto position

plt.grid(True, alpha=0.3)
plt.show()
````

## Annotations and Drawing on a Subplot

### Adding text with `text()` and `annotate()`

````python
# From the book example
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, 'b-', linewidth=2)

# Add text at specific position
ax.text(3, 0.8, 'Sine Wave', fontsize=14, fontweight='bold')
ax.text(8, 0.5, 'First peak', fontsize=12, color='red')

# Add annotation with arrow
ax.annotate('Local max', 
            xy=(np.pi/2, 1),           # Point to annotate
            xytext=(np.pi/2, 1.4),      # Text position
            arrowprops=dict(facecolor='black', shrink=0.05),
            fontsize=12)

ax.annotate('Local min',
            xy=(3*np.pi/2, -1),
            xytext=(3*np.pi/2, -1.4),
            arrowprops=dict(facecolor='black', shrink=0.05),
            fontsize=12)

plt.show()
````

### Plotting additional lines and shapes

````python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, 'b-', linewidth=2)

# Horizontal line
ax.axhline(y=0.5, color='r', linestyle='--', alpha=0.7)
ax.axhline(y=-0.5, color='r', linestyle='--', alpha=0.7)

# Vertical line
ax.axvline(x=5, color='g', linestyle=':', alpha=0.7)

# Rectangle shape
from matplotlib.patches import Rectangle
rect = Rectangle((1, 0.2), 2, 0.6, 
                 facecolor='yellow', alpha=0.3)
ax.add_patch(rect)

# Arrow
from matplotlib.patches import Arrow
arrow = Arrow(5, 0.8, 1, 0.3, width=0.2)
ax.add_patch(arrow)

plt.xlim(0, 10)
plt.ylim(-1.2, 1.2)
plt.show()
````

## Saving Plots to File

### Saving with `savefig()`

````python
# From the book example
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, 'b-', linewidth=2)
ax.set_title('Sine Wave')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')

# Save the figure
plt.savefig('sine_wave.png')          # PNG format
plt.savefig('sine_wave.jpg')          # JPEG format
plt.savefig('sine_wave.pdf')          # PDF format
plt.savefig('sine_wave.svg')          # SVG format

# With options
plt.savefig('sine_wave.png', 
            dpi=300,                   # Resolution
            bbox_inches='tight',       # Remove whitespace
            facecolor='white',         # Background color
            transparent=False)         # Transparent background

# Close the figure to free memory
plt.close('all')
````

#### **Summary Table - Matplotlib API**

| Concept | Description | Example |
|---------|-------------|---------|
| **Figure** | The entire window | `fig = plt.figure()` |
| **Axes** | A single plotting area | `ax = fig.add_subplot(2,2,1)` |
| **plot()** | Line plot | `ax.plot(x, y)` |
| **scatter()** | Scatter points | `ax.scatter(x, y)` |
| **hist()** | Histogram | `ax.hist(data, bins=20)` |
| **bar()** | Bar chart | `ax.bar(x, height)` |
| **set_title()** | Title | `ax.set_title('Title')` |
| **set_xlabel()** | X-axis label | `ax.set_xlabel('X')` |
| **legend()** | Legend | `ax.legend()` |
| **savefig()** | Save the figure | `plt.savefig('plot.png')` |


###  Complete example combining everything

````python
# Data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create figure with custom style
plt.style.use('seaborn-v0_8-darkgrid')
fig, ax = plt.subplots(figsize=(10, 6))

# Plot
ax.plot(x, y1, 'b-', linewidth=2, label='sin(x)')
ax.plot(x, y2, 'r--', linewidth=2, label='cos(x)')

# Customize
ax.set_title('Sine and Cosine Waves', fontsize=16, fontweight='bold')
ax.set_xlabel('X (radians)', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π'])
ax.legend(loc='upper right', fontsize=12)
ax.grid(True, alpha=0.3)

# Annotate
ax.annotate('sin(x) max', xy=(np.pi/2, 1), xytext=(2, 1.2),
            arrowprops=dict(facecolor='black', shrink=0.05))

# Save
plt.savefig('trig_functions.png', dpi=150, bbox_inches='tight')
plt.show()
````
