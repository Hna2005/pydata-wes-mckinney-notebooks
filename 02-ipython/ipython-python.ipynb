### Essential Keyboard Shortcuts

| Key | Meaning | Usage |
|-----|---------|-------|

### Table 2-1: IPython Keyboard Shortcuts

| Shortcut | Name | Description | Example |
|----------|------|-------------|---------|
| `Ctrl+P` / `↑` | Reverse search | Move backward in command history | `Ctrl+P` → previous command |
| `Ctrl+N` / `↓` | Forward search | Move forward in command history | `Ctrl+N` → next command |
| `Ctrl+R` | Advanced search | Search history by any substring | `Ctrl+R` + `np` → finds `np.mean()` |
| `Ctrl+Shift+V` | Paste | Paste from clipboard (Linux/WSL) | Paste copied text |
| `Ctrl+C` | Interrupt | Stop currently running code | Stops infinite loop |
| `Ctrl+A` | Line start | Move cursor to beginning of line | `\|print()` → `print()` |
| `Ctrl+E` | Line end | Move cursor to end of line | `print()\|` |
| `Ctrl+K` | Cut to end | Cut from cursor to end of line | `print(\|"Hi")` → `print(\|)` |
| `Ctrl+U` | Cut to start | Cut from cursor to start of line | Clears current line |
| `Ctrl+F` | Forward char | Move cursor one character forward | `Hel\|lo` → `Hel\|lo` |
| `Ctrl+B` | Backward char | Move cursor one character backward | `Hel\|lo` → `He\|llo` |
| `Ctrl+L` | Clear screen | Clear terminal display | History remains intact |

---

### Quick Tips

- **`Ctrl+R`** is the most powerful — searches *anywhere* in command, not just start
- **`Ctrl+K` vs `Ctrl+U`**: Cut forward vs cut backward from cursor
- **`Ctrl+C`** kills only the running code, not IPython itself
- **`Ctrl+L`** only clears view — your history is still there

## getattr & iterable

```python
def clean_data(data, operations):
    """اعمال لیستی از عملیات پاکسازی روی داده"""
    for op in operations:
        if hasattr(data, op): #امکان انجام عمل رو داره؟
            data = getattr(data, op)() #data.op()
    return data

text = "  Hello World!  "
ops = ['strip', 'lower', 'capitalize']
result = clean_data(text, ops)  # 'Hello world!'
```

```python
method_name = 'upper'
result = getattr(a, method_name)()  # معادل a.upper()
print(result)  # 'HELLO'

# بررسی وجود صفت:
if hasattr(a, 'lower'):
    print("می‌تواند lowercase کند")
```

در پایتون، هر چیزی که بشه رویش با for حلقه زد، iterable است.
(در اعداد امکان پذیر نیست)

```python
def isiterable(obj):
    try:
        iter(obj)
        return True
    except TypeError:
        return False

isiterable([1, 2, 3])    # True
isiterable("hello")      # True
isiterable(123)          # False
isiterable({'a': 1})     # True
```

## mutable objects

```python
# list
a_list = ["ali", 2, [4, 5]]
a_list[2] = (3, 4)   
# ['ali', 2, (3, 4)]

# dict
a_dict = {"name": "Ali", "aga" : 20}
a_dict["aga"] = 25
# {'name': 'Ali', 'aga': 25}

# set
a_set = {"a", 2, 3}
a_set.add(4)
# {'a', 2, 3, 4}

# numpy array
import numpy as np
arr = np.array([1, 2, 3])
arr[0] = 100
# [100   2   3]
```

### usage:

```python
def process_data(data):
    # dangerous
    data.append(100)
    return data

# change the original data

def process_data_safe(data):
    data = data.copy()
    data.append(100)
    return data
```

2. dictionary keys

```python
# just immutable
valid_dict = {
    'name': 'Ali',     # str - immutable
    1: 'one',          # int - immutable 
    (1, 2): 'tuple'    # tuple - immutable 
}

invalid_dict = {
    [1, 2]: 'list'     # TypeError: unhashable type: 'list'
}
```

3. is or not:

```python
# immutable
a = "hello"
b = "hello"
print( a is b) # true

# mutable
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)  # False 
```

## dates and times

```python
from datetime import datetime, date, time, timedelta
# year , month, day, hour, min, second
dt = datetime(2005, 10, 29, 20, 30, 21)
print(dt)   # 2005-10-29 20:30:21

print(dt.year)     # 2011
print(dt.month)    # 10
print(dt.day)      # 29
print(dt.hour)     # 20
print(dt.minute)   # 30
print(dt.second)   # 21
print(dt.weekday()) # 5

# export date and time
dt. date()  # 2005-10-29
dt.time()   # 20:30:21
```

### datetime to string

```python
dt.strftime('%Y/%m/%d %H:%M %A, %B, %d')
# '2005/10/29 20:30 Saturday, October, 29'
```

### string to datetime

```python
date_str = '20091031'
dt = datetime.strptime(date_str, '%Y%m%d')
datetime.strptime('29 Oct 2011', '%d %b %Y') # 2011-10-29 00:00:00
```

### timedelta 

```python
dt1 = datetime(2011, 10, 29, 20, 30, 21)
dt2 = datetime(2011, 11, 15, 22, 30)

delta = dt2 - dt1
print(delta)                    # 17 days, 1:59:39
print(delta.days)              # 17
print(delta.seconds)           # 7179 (1:59:39 به ثانیه)
print(delta.total_seconds())   # 1477179.0

from datetime import timedelta
tomorrow = datetime.now() + timedelta(days=1)
next_week = datetime.now() + timedelta(weeks=1)
one_hour_later = datetime.now() + timedelta(hours=1)
```
