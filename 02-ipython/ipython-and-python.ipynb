# ipython

## Installing IPython

### Windows (PowerShell)

```powershell
pip install ipython
```


IPython can be installed using `pip`, Python's package manager. This command downloads and installs the latest stable version of IPython into the current Python environment.

### Linux

```bash
pip3 install ipython
```

On Linux, `pip3` is commonly used to install packages for Python 3. After installation, IPython can be launched from the terminal.

## Launching IPython

After installing IPython, start an interactive session by running the following command in your terminal.

```bash
ipython
```

**Explanation:**

Running the `ipython` command launches the IPython interactive shell. You will see a prompt such as `In [1]:`, indicating that IPython is ready to execute Python commands interactively.

![Linching IPython](images/00-launching-ipython.png)

### Example

```python
In [1]: print("Hello, IPython!")
Hello, IPython!

In [2]: 2 ** 10
Out[2]: 1024
```

This example demonstrates how IPython executes Python statements interactively and displays the result immediately.

## IPython Help System: `?` vs `??`

IPython provides a powerful built-in help system for exploring objects interactively.

- `?` displays basic information such as the type, signature, and docstring.
- `??` provides extended information and, when available, the object's source code.

### Example 1: Inspecting an Object

Use `?` to quickly inspect an object's information.

```python
In [1]: name = "Ali"

In [2]: name?
Type:         str
String form:  Ali
Length:       3
Docstring:
str(object='') -> str
...
```

The `?` operator displays useful information about an object without leaving the IPython session.

![question mark](images/01-question-mark.png)


### Example 2: Inspecting a Custom Function

Create a function with a docstring and inspect it using both `?` and `??`.

```python
In [3]: def sum(a, b):
   ...:     """Sum of two numbers."""
   ...:     return a + b
```

![??](images/02-double-q-mark.png)

#### Using `??`

```python
In [4]: sum??

Signature: sum(a, b)
Source:
def sum(a, b):
    """Sum of two numbers."""
    return a + b

Type: function
```

`??` displays the function's source code in addition to its metadata, when the source is available.
#### Using `?`

```python
In [5]: sum?

Signature: sum(a, b)
Docstring: Sum of two numbers.
Type: function
```

Use `?` when you only need a quick overview of an object, such as its signature and documentation.

### When to use `??` over `?`
- `?` is enough for **built-in types** (str, int, list, etc.)
- `??` is essential for **custom functions** to verify logic
- `??` is also useful for **third-party libraries** to understand how they work

## Running Python Scripts: `%run` vs `%run -i`

The `%run` magic command executes an external Python script directly from an IPython session.

### Using `%run -i`

```python
In [1]: x = 10

In [2]: y = 5

In [3]: %run -i tmp.py
12
```

The `-i` option runs the script in the current IPython namespace. Variables already defined in the interactive session are available inside the script.

### Using `%run`

```python
In [4]: %run tmp.py

NameError: name 'x' is not defined
```

Without the `-i` option, the script runs in a fresh namespace. Variables from the current IPython session are **not** accessible unless they are defined inside the script itself.

## Loading a Script with `%load`

The `%load` magic inserts the contents of a Python file into the current IPython input cell instead of executing it.

```python
In [5]: %load tmp.py

In [6]:
# %load tmp.py
y = 2
result = x + y
print(result)
```

This is useful when you want to review or edit a script before running it.

![run](images/04-run.png)

**Use `%run -i` when your script needs variables from the current session.** The `-i` flag shares the notebook's namespace with the script. Without it, the script runs in isolation and cannot access existing variables, causing `NameError`.

### Summary

| Command | Description |
|---------|-------------|
| `%run script.py` | Executes a script in a new namespace. |
| `%run -i script.py` | Executes a script using the current IPython namespace. |
| `%load script.py` | Loads the script into the current input cell without executing it. |

## Measuring Execution Time: `%time` and `%timeit`

IPython provides built-in magic commands for measuring code execution time.

### Using `%timeit`

```python
In [17]: %timeit sum(range(1000))

7.21 µs ± 59.7 ns per loop
(mean ± std. dev. of 7 runs, 100,000 loops each)
```

`%timeit` executes the code multiple times and reports the average execution time. It is the preferred tool for benchmarking small code snippets.
### Using `%time`

```python
In [18]: %time sum(range(1000))

CPU times: user 19 µs, sys: 0 ns, total: 19 µs
Wall time: 21.5 µs

Out[18]: 499500
```

`%time` executes the code only once and reports both CPU time and wall-clock time. It is useful for measuring a single execution.

![time](images/07-time.png)

## Pasting Code: `%paste` and `%cpaste`

IPython provides two magic commands for pasting multi-line code directly into the interactive session.

### Using `%paste`

```python

y = 8

```

`%paste` inserts code directly from the clipboard and executes it immediately. Any indentation errors in the pasted code will be reported.

### Using `%cpaste`

```python

Pasting code; enter '--' alone on the line to stop or use Ctrl-D.
```

`%cpaste` enters a special paste mode, allowing you to type or paste multi-line code manually. Finish by entering `--` on a new line (or press `Ctrl+D` on Unix-like systems).

![paste](images/08-paste.png)

## Copying Objects

Python provides two common ways to copy objects: shallow copy and deep copy. The appropriate choice depends on whether the object contains nested mutable objects.

````python
# shallow copy
a = [1, 2, 3, 4]
b = a.copy()
b = a[:]
b = list(a)

#deep copy
import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
````

## Dynamic Attribute Access

Python allows attributes and methods to be accessed dynamically using `getattr()` and checked using `hasattr()`. This is especially useful when attribute names are determined at runtime.

````python
def clean_data(data, operations):
    """Apply a sequence of cleaning operations to the input data."""
    for op in operations:
        # Check whether the object has the specified method
        if hasattr(data, op):
            # Dynamically call the method
            data = getattr(data, op)()
    return data
````

````python
text = "  Hello World!  "

# List of methods to apply
ops = ["strip", "lower", "capitalize"]
result = clean_data(text, ops)
print(result)  # Hello world!
````

````python
# Dynamically call a method by its name
method_name = "upper"
result = getattr(text, method_name)()  # Equivalent to text.upper()
print(result)  # HELLO WORLD!

# Check whether an object supports a specific method
if hasattr(text, "lower"):
    print("The object supports lowercase conversion.")
````

## Iterable Objects

An **iterable** is any object that can be traversed using a `for` loop. Common iterable objects include strings, lists, tuples, dictionaries, and sets.

````python
def isiterable(obj):
    try:
        iter(obj)
        return True
    except TypeError:
        return False

isiterable([1, 2, 3])    # True
isiterable("hello")      # True
isiterable(123)          # False
isiterable({'a': 1})     # True
````

## Mutable Objects

Mutable objects can be modified after they are created.## mutable objects

````python
# List
a_list = ["ali", 2, [4, 5]]
a_list[2] = (3, 4)   
# ['ali', 2, (3, 4)]

# Dictionary
a_dict = {"name": "Ali", "aga" : 20}
a_dict["aga"] = 25
# {'name': 'Ali', 'aga': 25}

# set
a_set = {"a", 2, 3}
a_set.add(4)
# {'a', 2, 3, 4}

# NumPy Array
import numpy as np
arr = np.array([1, 2, 3])
arr[0] = 100  # Modify an element
# [100   2   3]
````

## Copying for Data Safety

When working with mutable objects, modifying a function argument directly may also modify the original object. Creating a copy helps prevent unintended side effects.

````python
def process_data(data):
    # Modify the original list
    data.append(100)
    return data
````

````python
def process_data_safe(data):
    # Work on a copy instead of the original list
    data = data.copy()
    data.append(100)
    return data
````

## Object Identity: `is`

The `is` operator checks whether two variables reference the **same object** in memory, not whether they have the same value.

````python
# Immutable objects

a = "hello"
b = "hello"

print(a is b)

# Mutable objects

a = [1, 2, 3]
b = [1, 2, 3]

print(a is b)
````

## Boolean Short-Circuit Evaluation

````python
def expensive_function():
    return True

# The function is not called because the result is already False
print(False and expensive_function())

# The function is not called because the result is already True
print(True or expensive_function())
````

## Working with Dates and Times

````python
from datetime import datetime, date, time, timedelta

# Create a datetime object
dt = datetime(2005, 10, 29, 20, 30, 21)
print(dt)
````

````python
# Access individual components

print(dt.year)
print(dt.month)
print(dt.day)

print(dt.hour)
print(dt.minute)
print(dt.second)

print(dt.weekday())
````

````python
# Extract the date and time

print(dt.date())
print(dt.time())
````

## Converting Datetime to a String

````python
formatted = dt.strftime("%Y/%m/%d %H:%M %A, %B %d")
print(formatted)
````

## Parsing Strings into Datetime Objects

````python
date_str = "20091031"

dt = datetime.strptime(date_str, "%Y%m%d")
print(dt)
````

````python
print(datetime.strptime("29 Oct 2011", "%d %b %Y"))
````

## Time Differences with `timedelta`

````python
dt1 = datetime(2011, 10, 29, 20, 30, 21)
dt2 = datetime(2011, 11, 15, 22, 30)

delta = dt2 - dt1

print(delta)
print(delta.days)
print(delta.seconds)
print(delta.total_seconds())
````

````python
# Create future dates using timedelta

tomorrow = datetime.now() + timedelta(days=1)
next_week = datetime.now() + timedelta(weeks=1)
one_hour_later = datetime.now() + timedelta(hours=1)

print(tomorrow)
print(next_week)
print(one_hour_later)
````
