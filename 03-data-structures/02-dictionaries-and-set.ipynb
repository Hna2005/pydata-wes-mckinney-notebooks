### `setdefault()`

**Key points:**
- If key exists: returns the existing value
- If key doesn't exist: inserts key with default value ([]), then returns that value

```python
people = [('ali', 25), ('sara', 30), ('reza', 25), ('maryam', 30)]
by_age = {}

for name, age in people:
    by_age.setdefault(age, []).append(name)
by_age
```

### `defaultdict`

```python
from collections import defaultdict

# defaultdict takes a factory function
by_letter = defaultdict(list)  # default empty list
words = ['apple', 'bat', 'bar', 'atom', 'book']
for word in words:
    by_letter[word[0]].append(word)  # no need for setdefault!

print(by_letter)
print(dict(by_letter))  # Convert to normal dict
```

```python
# Other default factories
int_dict = defaultdict(int)  # default 0
int_dict['a'] += 1
print(int_dict['a'])  # 1
print(int_dict['b'])  # 0 (automatically created)
```

```python
float_dict = defaultdict(float)  # default 0.0
str_dict = defaultdict(str)      # default ''
set_dict = defaultdict(set)      # default set()
```

### Hashable Objects

This function checks if an object can be used as a dictionary key. In Python, dictionary keys must be hashable (immutable).

```python
def is_hashable(obj):
    try:
        hash(obj)
        return True
    except TypeError:
        return False

print(is_hashable(42))        # True
print(is_hashable('hello'))   # True
print(is_hashable((1, 2)))    # True
print(is_hashable([1, 2]))    # False
print(is_hashable({'a': 1}))  # False
```

### Dictionary Merge

```python
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
dict3 = {'e': 5, 'f': 6}

# Using | operator (Python 3.9+)
merged = dict1 | dict2 | dict3
print(merged) # create a new dict

# In-place update with |=
dict1 |= dict2
print(dict1)  # update dict
```

```python
# | is equivalent to:
merged = {**dict1, **dict2, **dict3}

# |= is equivalent to:
dict1.update(dict2)
```

### Groupby:
(defaultdict)

```python
# Group people by age
people = [
    {'name': 'Ali', 'age': 25},
    {'name': 'Sara', 'age': 30},
    {'name': 'Reza', 'age': 25},
    {'name': 'Maryam', 'age': 30},
    {'name': 'Hassan', 'age': 35}
]

groups = defaultdict(list)
for person in people:
    groups[person['age']].append(person['name'])

print(dict(groups))
# {25: ['Ali', 'Reza'], 30: ['Sara', 'Maryam'], 35: ['Hassan']}

# or
by_age = {}
for name, age in people:
    by_age.setdefault(age, []).append(name)
# {25: ['ali', 'reza'], 30: ['sara', 'maryam']}
```

| Method | Description | Example | Result |

### Time and Memory

```python
# List search: O(n)
def list_search():
    lst = list(range(1000))
    return 999 in lst

# Dict search: O(1)
def dict_search():
    d = {i: i for i in range(1000)}
    return 999 in d

# Dictionary is MUCH faster for lookups
# list_search: ~10 µs
# dict_search: ~0.1 µs
```

```python
import sys

# Dictionary has overhead
d = {i: i for i in range(1000)}
print(sys.getsizeof(d))  # ~50KB

# List uses less memory for sequential data
lst = list(range(1000))
print(sys.getsizeof(lst))  # ~8KB

# Trade-off: memory vs speed
# Dictionary has overhead
d = {i: i for i in range(1000)}
print(sys.getsizeof(d))  # ~50KB

# List uses less memory for sequential data
lst = list(range(1000))
print(sys.getsizeof(lst))  # ~8KB

# Trade-off: memory vs speed
```

### Set Operations

```python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7, 8}
c = {5, 6, 7, 8, 9}

# Union: elements in either set
print(a.union(b)) 
print(a | b | c)
```

```python
# Intersection: elements in both sets
print(a.intersection(b)) 
print(a & b & c)
```

```python
# Difference: elements in a but not in b
print(a.difference(b)) 
print(b - a) 
```

```python
# Symmetric difference: elements in either set but not both
print(a.symmetric_difference(b))
print(a ^ b)
print((a - b) | (b - a)) 
```

### In-Place

```python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7}

# In-place union
a |= b  # same as a.update(b)
print(a)
```

```python
# In-place intersection
a = {1, 2, 3, 4, 5}
a &= b
print(a)
```

```python
# In-place difference
a = {1, 2, 3, 4, 5}
a -= b
print(a)
```

```python
# In-place symmetric difference
a = {1, 2, 3, 4, 5}
a ^= b
print(a)
```

### Subset & Superset

```python
a = {1, 2, 3}
b = {1, 2, 3, 4, 5}
c = {1, 2, 3}

# issubset() - all elements of a are in b
print(a.issubset(b)) 
print(a <= b) 
```

```python
# Proper subset (strict)
print(a < b)           # True (a is proper subset of b)
print(a < c)           # False (equal, not proper)
```

```python
# issuperset() - all elements of b are in a
print(a.issuperset(b)) 
print(a >= b)
```

```python
# disjooint()
print(a.isdisjoint(b)) # False (1, 2, 3 are common)
```

## `frozenset()`

```python
# frozenset is immutable version of set
fs = frozenset([1, 2, 3, 2, 1])
print(fs) 
print(type(fs))
```

```python
# Cannot modify
# fs.add(4)  # AttributeError!
# fs.remove(1)  # AttributeError!

# But can use in set operations
other = {2, 3, 4}
print(fs | other) 
print(fs & other) 
```

### Usage

```python
# Can be used as dictionary key (unlike set)
valid_dict = {
    frozenset([1, 2]): "value 1",
    frozenset([3, 4]): "value 2"
}
print(valid_dict[frozenset([1, 2])])  # "value 1"
```

```python
# Can be element of another set (unlike set)
set_of_sets = {frozenset([1, 2]), frozenset([3, 4])}
print(set_of_sets) 

# This would fail with normal sets
# set_of_sets = {{1, 2}, {3, 4}}  # TypeError!
```

## List Comprehemsion

 Basic syntax:


`[expression for value in collection if condition]`

```python
numbers = [1, 2, 3, 4, 5]

squares = [x ** 2 for x in range(10)]
print("squares:", squares)

even_squares = [x ** 2 for x in numbers if x % 2 == 0]
print("even_squares:", even_squares)

transformed = [x * 2 if x % 2 == 0 else x ** 2 for x in numbers]
print("transformed:", transformed)
```

```python
# nested loops
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [item for row in matrix for item in row]
flattened
```

```python
# Cartesian product
colors = ['red', 'green']
sizes = ['S', 'M', 'L']

# All combinations
combinations = [(color, size) for color in colors for size in sizes]
print(combinations)
```

```python
combinations = [(color, size) for color in colors for size in sizes if color != 'green' or size != 'L']
print(combinations)

transposed = [[row[i] for row in matrix] for i in range(3)]
print(transposed)
```

### Dict Comprehension

**Syntax:**

 `{key_expression: value_expression for value in collection if condition}`

```python
strings = ['a', 'as', 'bat', 'car', 'dove', 'python']
loc_mapping = {val: index for index, val in enumerate(strings)}
loc_mapping
```

```python
data = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

# Filter by key
filtered_by_key = {k: v for k, v in data.items() if k > 'c'}
print(filtered_by_key)
```

```python
# Filter by value
filtered_by_value = {k: v for k, v in data.items() if v % 2 == 0}
print(filtered_by_value) 
```

```python
# Transform values
transformed = {k: v ** 2 for k, v in data.items() if v > 2}
print(transformed)
```

```python
some_tuples = [(1, 2, 3), (4, 5, 6), (7, 8, 9)]

# Nested comprehension (flatten)
flattened = [x for tup in some_tuples for x in tup]
print(flattened)
```

```python
# Comprehension inside comprehension (matrix)
matrix = [[x for x in tup] for tup in some_tuples]
print(matrix) 
# The difference:
# - First: one list with all elements
# - Second: list of lists (preserves structure)
```

```python
# Another example: transpose
transposed = [[row[i] for row in matrix] for i in range(3)]
print(transposed)
```

| Type | Syntax | Output | Example |
|------|--------|--------|---------|
| **List** | `[expr for item in iter if cond]` | List | `[x**2 for x in range(5)]` |
| **Set** | `{expr for item in iter if cond}` | Set | `{x**2 for x in range(5)}` |
| **Dict** | `{k_expr: v_expr for item in iter if cond}` | Dictionary | `{x: x**2 for x in range(5)}` |
| **Nested List** | `[expr for sub in outer for item in sub]` | Flattened List | `[x for row in mat for x in row]` |
