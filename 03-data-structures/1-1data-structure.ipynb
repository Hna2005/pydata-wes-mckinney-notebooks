## Tuple

```python
# create tuple
tup = 1, 2, 3
print(tup) # (1, 2, 3)

tup = (1, 2, 3)
print(tup) # (1, 2, 3)

single_without_comma = (42)      # it's int
single_with_comma = (42,)        # it's tuple

# nested tuple
nasted_tup = (1, 2, 3), (4, 5)
len(nasted_tup)    # 3
len(nasted_tup[0]) # 2
nasted_tup[1][0]   # 4 
```

### replication

```python
# creates a new tuple with repeated elements
tup = ('foo', 'bar')
result = tup * 4
print(result)  
# ('foo', 'bar', 'foo', 'bar', 'foo', 'bar', 'foo', 'bar')

# Objects are not copied, only references are repeated
nested = ([1, 2],) * 3
print(nested)  # ([1, 2], [1, 2], [1, 2])
nested[0].append(3)  # modify one...
print(nested)  # ([1, 2, 3], [1, 2, 3], [1, 2, 3])  # all changed!
```

### unpacking 

```python
values = 1, 2, 3, 4, 5
a, b, c, d, e = values
print(a)

# unpacking nested tuple
tup = 4, 5, (6, 7)
m, n, (t, z) = tup

# Extended unpacking with * operator
a, b, *rest = values
print(a)       # 1
print(b)       # 2
print(rest)    # [3, 4, 5]  (note: becomes a list!)

# Can be used in the middle as well
a, *middle, b = values
print(a)       # 1
print(middle)  # [2, 3, 4]
print(b)       # 5

# * can only appear once
# a, *mid, *more, b = values  # SyntaxError!
```

```python
# Using underscore (_) for unused values

values = 1, 2, 3, 4, 5

# If we only want the first two elements
a, b, *_ = values
print(a)  # 1
print(b)  # 2
# rest are ignored, so we use _

# Or if we only want the first element
first, *_ = values
print(first)  # 1
```

### add and delete

```python
b_list = ['foo', 'peekaboo', 'baz']

# add to the end
b_list.append('dwarf')
b_list.append(['x', 'y']) # ['foo', 'peekaboo', 'baz', 'dwarf', ['x', 'y']]

# add to specified position
b_list.insert(1, 'red') 
b_list.insert(-1, 'blue') # ['foo', 'red', 'peekaboo', 'baz', 'blue', 'dwarf']
```

```python
from collections import deque
d = deque(['foo', 'bar'])
d.appendleft('start')  # O(1) - faster than insert!
```

```python
# Removing Items with pop()
b_list = ['foo', 'red', 'peekaboo', 'baz', 'dwarf']

# Without argument: removes and returns last element
item = b_list.pop()
print(item)     # 'dwarf'
print(b_list)   # ['foo', 'red', 'peekaboo', 'baz']

# With index: removes and returns element at that index
item = b_list.pop(1)
print(b_list)   # ['foo', 'peekaboo', 'baz']

# pop from beginning is expensive (other elements must shift)
# first = b_list.pop(0)  # O(n) - slow!
```

```python
# Removing Specific Items with remove()
b_list = ['foo', 'red', 'baz', 'dwarf', 'foo']

# Removes the first matching occurrence
b_list.remove('foo')
print(b_list)  # ['red', 'baz', 'dwarf', 'foo']  # First 'foo' removed

# If element doesn't exist: ValueError

# Removing all matching items:
while 'foo' in b_list:
    b_list.remove('foo')
print(b_list)  # ['red', 'baz', 'dwarf']
```

```python
# Checking for Item Existence in List
b_list = ['red', 'baz', 'dwarf']

print('dwarf' in b_list)      # True
print('blue' in b_list)       # False
print('dwarf' not in b_list)  # False

# Note: List search is O(n)
# For repeated lookups, use a set instead
```

### extend and append

```python
# Difference between extend and append
x = [1, 2, 3]
x.append([4, 5])   # adds the list itself as a single element
# [1, 2, 3, [4, 5]]

x = [1, 2, 3]
x.extend([4, 5])   # adds the elements of the list individually
print(x)  # [1, 2, 3, 4, 5]
```

### sort()

```python
a = [7, 2, 5, 1, 3]
a.sort() # [1, 2, 3, 5, 7]
a.sort(reverse=True) # [7, 5, 3, 2, 1]
```

```python
# Sorting lists with key parameter
b = ['saw', 'small', 'He', 'foxes', 'six']

# Sort by word length
b.sort(key=len)
# ['He', 'saw', 'six', 'small', 'foxes']

# Sort by last character of each word
b.sort(key=lambda x: x[-1])
print(b)
# ['He', 'small', 'foxes', 'saw', 'six'] depends on implementation

# Case-sensitive vs case-insensitive sorting
words = ['banana', 'Apple', 'cherry', 'Date']

# Default sort (ASCII order): uppercase comes before lowercase
words.sort()  # ['Apple', 'Date', 'banana', 'cherry']

# Case-insensitive sort (lexicographic order)
words.sort(key=str.lower)  # ['Apple', 'banana', 'cherry', 'Date']
```

### sorted()

```python
a = [7, 2, 5, 1, 3]
b = sorted(a)
print(a)  # [7, 2, 5, 1, 3]
print(b)  # [1, 2, 3, 5, 7]

sorted_set = sorted({3, 1, 2})  # [1, 2, 3]
sorted_dict = sorted({'b': 2, 'a': 1, 'c': 3})  # ['a', 'b', 'c']
```

| Feature | `sorted()` | `sort()` |
|---------|------------|----------|
| **In-place modification** | No - creates a new list | Yes - modifies the original list |
| **Applicability** | Works on any iterable | Works only on lists |
| **Memory** | More (creates new list) | Less (uses same list) |

### bisect

```python
import bisect
c = [1, 2, 2, 2, 3, 4, 7]

# bisect: returns the index where to insert item to keep list sorted
print(bisect.bisect(c, 2))   # 4 (after the last 2)
print(bisect.bisect(c, 5))   # 6 (between 4 and 7)

c = [1, 2, 2, 2, 3, 4, 7]

bisect.insort(c, 6)
print(c)  # [1, 2, 2, 2, 3, 4, 6, 7]
```

## Built-in Sequence Functions

### enumerate

```python
collection = ['foo', 'bar', 'baz']
for i, value in enumerate(collection):
    print(f"{i}: {value}")
# 0: foo
# 1: bar
# 2: baz

collection = ['foo', 'bar', 'baz']
enum = enumerate(collection)
print(enum)           # <enumerate object at 0x...>
print(list(enum))     # [(0, 'foo'), (1, 'bar'), (2, 'baz')]

# start from 1
for i, value in enumerate(collection, start=1):
    print(f"{i}: {value}")

# mapping dictionary
mapping2 = {v: i for i, v in enumerate(collection)}
print(mapping2)  # {'foo': 0, 'bar': 1, 'baz': 2}

# in matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

def find_coordinate(matrix, target):
    for i, row in enumerate(matrix):
        for j, val in enumerate(row):
            if val == target:
                return (i, j)
    return None

print(find_coordinate(matrix, 5))  # (1, 1)
```

### Sorting complex data

```python
data = [
    {'name': 'Ali', 'age': 30, 'score': 85},
    {'name': 'Sara', 'age': 25, 'score': 92},
    {'name': 'Reza', 'age': 35, 'score': 78},
    {'name': 'Maryam', 'age': 28, 'score': 92}
]

# Sort by age
by_age = sorted(data, key=lambda x: x['age'])

# Sort by score (descending) and then by name
by_score_name = sorted(data, 
                      key=lambda x: (-x['score'], x['name']))

for item in by_score_name:
    print(f"{item['name']}: {item['score']}")
# Maryam: 92
# Sara: 92
# Ali: 85
# Reza: 78
```

### zip 

```python
seq1 = ['foo', 'bar', 'baz']
seq2 = ['one', 'two', 'three']

zipped = zip(seq1, seq2)
print(zipped)        # <zip object at 0x...>
print(list(zipped))  # [('foo', 'one'), ('bar', 'two'), ('baz', 'three')]
```

#### zip with unequal sequences

```python
seq1 = ['foo', 'bar', 'baz']
seq2 = ['one', 'two']

# Shortest sequence determines the length
zipped = zip(seq1, seq2)
print(list(zipped))  # [('foo', 'one'), ('bar', 'two')]
# 'baz' has no pair, so it's ignored

# If you want the longest sequence to be the basis, use itertools.zip_longest
from itertools import zip_longest
zipped = zip_longest(seq1, seq2, fillvalue='MISSING')
print(list(zipped))
# [('foo', 'one'), ('bar', 'two'), ('baz', 'MISSING')]
```

```python
# Combining with enumerate:
seq1 = ['foo', 'bar', 'baz']
seq2 = ['one', 'two', 'three']

for i, (a, b) in enumerate(zip(seq1, seq2)):
    print(f"{i}: {a} - {b}")
# 0: foo - one
# 1: bar - two
# 2: baz - three
```

### reversed 

```python
a = [1, 2, 3, 4, 5]

# reversed: returns a generator (memory efficient)
rev = reversed(a)
print(rev)          # <list_reverseiterator object>
print(list(rev))    # [5, 4, 3, 2, 1]

# [::-1]: creates a new list (more memory)
rev2 = a[::-1]
print(rev2)         # [5, 4, 3, 2, 1]
print(type(rev2))   # <class 'list'>
```

```python
# Working with tuples
t = (1, 2, 3)
print(tuple(reversed(t)))  # (3, 2, 1)

# Working with strings
s = "python"
print(''.join(reversed(s)))  # 'nohtyp'

# Working with range
r = range(5)
print(list(reversed(r)))     # [4, 3, 2, 1, 0]

# Working with dict (keys)
d = {'a': 1, 'b': 2, 'c': 3}
print(list(reversed(d)))      # ['c', 'b', 'a']
```

### Access

### add and change

```python
d1 = {'a': 'some value', 'b': [1, 2, 3, 4], 7: 'an integer'}

# keys:
print(d1['b'])  # [1, 2, 3, 4]
print(d1.keys()) # dict_keys(['a', 'b', 7])
print(list(d1.keys()))  # ['a', 'b', 7]

# values:
d1.values()
list(d1.values())

# key-value pairs
d1.items()
list(d1.items())

# Can use in loops
for key in d1.keys():  # same as 'for key in d1:'
    print(key)

for value in d1.values():
    print(value)

for key, value in d1.items():
    print(f"{key}: {value}")
```

### delete

```python
# del
del d1[7]
# {'a': 'some value', 'b': [1, 2, 3, 4]}
# del doesn't return the value

# pop
ret = d1.pop('b')
print(ret) # [1, 2, 3, 4]
ret = d1.pop('nonexistent', 'default')
# If key exists: returns its value and removes the key-value pair
# If key doesn't exist: returns the default value (no error!)

key, value = d1.popitem()
# removes last inserted item

d1.clear()
# {}
```

### update

```python
d1 = {'a': 'some value', 'b': [1, 2, 3, 4], 7: 'an integer'}

# Update with another dict
d1.update({'b': 'foo', 'c': 12})
print(d1)  # {'a': 'some value', 'b': 'foo', 7: 'an integer', 'c': 12}

# Update with list of tuples
d1.update([('x', 100), ('y', 200)])
print(d1)  # ... 'x': 100, 'y': 200

# Update with keyword arguments
d1.update(z=300, w=400)
print(d1)  # ... 'z': 300, 'w': 400
```

### get

```python
# Safe access - returns None if key doesn't exist

print(d1.get('a'))  # 'some value'
print(d1.get('c', 'NOT FOUND'))  # 'NOT FOUND'
```
