## Series 

```python
# Create a Series from list
val = [1, 2, 3]
obj = pd.Series(val)

# Creating Series with string index
ind = ['a', 'b', 'c']
obj2 = pd.Series(val, index=ind)

# Creating Series from dictionary
sdata = {'Ohio': 35000, 'Texas': 71000, 'Oregon': 16000, 'Utah': 5000}
obj3 = pd.Series(sdata)

# Access and modify
obj['0'] = 5
obj2['a'] # 1
obj['0']  # 5

# Selecting multiple values
obj[[1, 2]]

# index order
states = ['California', 'Ohio', 'Oregon', 'Texas']
obj4 = pd.Series(sdata, index=states)

# Detecting missing data
pd.isnull(obj4)    # True for NaN value
pd.notnull(obj4)   # False for NaN value
obj4.isnull()      # instance method

obj4.name = 'population'      # Series name
obj4.index.name = 'state'     # index name

print(f'obj:\n{obj}')
print(f'\nobj2:\n{obj2}')
print(f'\nobj3:\n{obj3}')
print(f'\nobj4:\n{obj4}')
```

|Method |	Purpose |	Syntax Example	| Notes|
|-------|-----------|-------------------|------|
`df[val]`	|Select column(s) or filter rows|	`df['col'] or df[['col1','col2']]`	|Also accepts boolean arrays (filter rows), slices (slice rows), or boolean DataFrames
`df.loc[val]`|	Select rows by label	|`df.loc['row_label']`|	Label-based row selection
`df.loc[:, val]`	|Select columns by label|	`df.loc[:, 'col']`|	Label-based column selection
`df.loc[val1, val2]`|	Select rows & columns by label	|`df.loc['row', 'col']`	|Label-based 2D selection
`df.iloc[where]`|	Select rows by integer position	|`df.iloc[0]	`|Integer position-based row selection
`df.iloc[:, where]`|	Select columns by integer position	|`df.iloc[:, 0]`	|Integer position-based column selection
`df.iloc[where_i, where_j]`	|Select rows & columns by integer position|	`df.iloc[0, 1]`|	Integer position-based 2D selection
`df.at[label_i, label_j]`	| Get/set single value by label	| `df.at['row', 'col'] = 5	`| Fast scalar access by label
`df.iat[i, j]`|	Get/set single value by position|	`df.iat[0, 1] = 5`|	Fast scalar access by integer position
`reindex()`	| Select rows/columns by labels| 	`df.reindex(new_labels)`	|Align data with new index
`get_value()`/`set_value()`	|Get/set single value by label	|` df.get_value('row','col')`	|Legacy method (use `.at `instead)

## Pandas DataFrame Cheat Sheet



### **Creating DataFrames**
| Method | Example | Description |
|--------|---------|-------------|
| From dict of lists | `pd.DataFrame({'col1': [1,2], 'col2': [3,4]})` | Most common method |
| From dict of dicts | `pd.DataFrame({'A': {0:1, 1:2}, 'B': {0:3, 1:4}})` | Outer keys = columns, inner keys = index |
| From list of dicts | `pd.DataFrame([{'a':1, 'b':2}, {'a':3, 'b':4}])` | Each dict = row |
| From 2D array | `pd.DataFrame(np.array([[1,2],[3,4]]))` | With optional index/columns |


### **Accessing Data**
| Operation | Syntax | Returns |
|-----------|--------|---------|
| Single column | `df['col']` or `df.col` | Series |
| Single row | `df.loc['index']` | Series |
| Multiple columns | `df[['col1', 'col2']]` | DataFrame |
| Multiple rows | `df.loc[['idx1', 'idx2']]` | DataFrame |


### **Modifying DataFrames**
| Operation | Method |
|-----------|--------|
| Add column (Length must match)| `df['new_col'] = values` |  
| Delete column | `del df['col']` or `df.drop('col', axis=1)` |
| Rename columns | `df.rename(columns={'old':'new'})` | 
| Set index | `df.set_index('col')` | 
| Reset index | `df.reset_index()` | 


### **Useful Attributes/Methods**
| Attribute/Method | Description |
|------------------|-------------|
| `df.head(n)` | First n rows |
| `df.tail(n)` | Last n rows |
| `df.shape` | (rows, columns) |
| `df.columns` | Column names |
| `df.index` | Index values |
| `df.dtypes` | Data types per column |
| `df.values` | Underlying numpy array |
| `df.T` | Transpose |
| `df.describe()` | Summary statistics |


#### Important Notes
1. **Column access**: `df['col']` works for all names, `df.col` only for valid Python identifiers
2. **Views vs copies**: Column access returns a view (modifications affect original)
3. **Alignment**: Operations align on index
4. **Missing data**: Represented as NaN
5. **Display**: In Jupyter, shows as HTML table

### Arithmetic and data alignment

```python
s1 = pd.Series([7.3, -2.5, 3.4, 1.5], index=['a', 'c', 'd', 'e'])
s2 = pd.Series([-2.1, 3.6, -1.5, 4, 3.1], index=['a', 'c', 'e', 'f', 'g'])

print(s1.add(s2)) 
# value with same lable are added
# produces NaN for non-overlapping indices

print(s1.add(s2, fill_value=0))  
# Addition with fill_value (fills missing values before operation)
# s1: f=0, e=0 s2:d=0

s1 + 10 # adds 10 to all element
```

```python
df1 = pd.DataFrame(np.arange(12.).reshape((3, 4)), columns=list('abcd'))
df2 = pd.DataFrame(np.arange(20.).reshape((4, 5)), columns=list('abcde'))
df2.loc[1, 'b'] = np.nan
df1.add(df2, fill_value=0)
```

|Method|	Operator| Equivalent|	Purpose	Example|
|------|------------|-----------|------------------|
`add()` / `radd()`	|+|	Addition	|`df.add(other)` or` df + other`
`sub()` / `rsub()`	|-|	Subtraction|	`df.sub(other)` or `df - other`
`div()` / `rdiv()`|	/|	Division|	`df.div(other)` or `df / other`
`floordiv()` / `rfloordiv()`	|//	| Floor division|	`df.floordiv(other)` or` df // other`
`mul() / rmul()`|	*|	Multiplication	|`df.mul(other)` or `df * other`
`pow() / rpow()`|	**|	Exponentiation	|`df.pow(2)` or `df ** 2`

    Note: r prefix methods handle reverse operations (e.g., radd() for other + df)

### Function Application and Mapping

```python
# apply
df = pd.DataFrame(np.random.randint(-10, 10, size=(4, 3)), columns=list('abc'),
                  index=['Utah', 'Ohio', 'Texas', 'Oregon'])

format = lambda x: '%.2f' % x

print(df.map(format))
df['a'].map(format)
```

```python
f = lambda x: x.max() - x.min()

print(df.apply(f))
df.apply(f, axis='columns')
```

```python
# apply on Series
s_numeric = pd.Series([1, 4, 9, 16, 25])

# Using with additional parameters
def power(x, exponent=2):
    return x ** exponent

s_numeric.apply(power, exponent=3)
```

```python
# apply on DataFrame
df = pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': [10, 20, 30, 40],
    'C': [100, 200, 300, 400]
}, index=['row1', 'row2', 'row3', 'row4'])

print(df)

# Example 1: Applying on columns (axis=0 or axis='index') - default
column_sums = df.apply(np.sum)
print(column_sums)
```

```python
# Example 2: Applying on rows (axis=1 or axis='columns')
row_sums = df.apply(np.sum, axis=1)
print(row_sums)
```

```python
# Example 3: Applying custom function on each column
def column_stats(column):
    return pd.Series({
        'mean': column.mean(),
        'std': column.std(),
        'min': column.min(),
        'max': column.max()
    })

stats_df = df.apply(column_stats)
print(stats_df)
```

```python
# Example 4: Applying function that returns multiple values
def normalize_and_round(col):
    normalized = (col - col.min()) / (col.max() - col.min())
    return normalized.round(2)

normalized_df = df.apply(normalize_and_round)
print(normalized_df)
```

```python
# Example 5: Applying on rows with access to column names
def row_calculation(row):
    return row['A'] * 2 + row['B'] / 10 - row['C'] * 0.01

row_results = df.apply(row_calculation, axis=1)
print(row_results)
```

```python
# Creating sample DataFrame
df_numbers = pd.DataFrame({
    'X': [1.234, 2.567, 3.891],
    'Y': [4.123, 5.456, 6.789],
    'Z': [7.012, 8.345, 9.678]
})

print(df_numbers)

# Example 1: Applying function on each element
rounded_df = df_numbers.map(lambda x: round(x, 1))
print(rounded_df)
```

```python
# Example 2: Converting data types
df_strings = pd.DataFrame({
    'A': ['1', '2', '3'],
    'B': ['4.5', '5.6', '6.7'],
    'C': ['7', '8', '9']
})

converted_df = df_strings.map(lambda x: float(x) if '.' in x else int(x))
print("\nDataFrame:")
print(converted_df)
print(converted_df.dtypes)
```

```python
# Example 3: Applying conditional function on each element
def format_number(x):
    if x > 5:
        return f"**{x:.2f}**"
    else:
        return f"{x:.2f}"

formatted_df = df_numbers.map(format_number)
print("\nDataFrame :")
print(formatted_df)
```
