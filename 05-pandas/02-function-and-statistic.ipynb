## Sorting and Ranking

```python
data = {
    'Name': ['Ali', 'Sara', 'Reza', 'Zahra'],
    'Score': [85, 92, 78, 90]
}

df = pd.DataFrame(data)
df.sort_values(by='Score', ascending=False)
# defult:ascending , ascending=Falce : descending

df.sort_values(by=['Name', 'Score'])
# frist sort by name then sort by score

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Reza'],
    'Score': [85, 92, 78]
}, index=['c', 'a', 'b'])

sorted_df = df.sort_index()
# sort by index

df['Rank'] = df['Score'].rank()
# ranks from one through the number of valid data points in an array

df['Rank_Dense'] = df['Score'].rank(method='dense', ascending=False)
# 
df.rank(method='first')
df.rank(ascending=False, method='max')
df
```

```python
# Example data
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Score': [85, 92, 78, 92, 88],
        'Age': [25, 30, 22, 30, 28]}
df = pd.DataFrame(data)

# 1. Using rank()
df['Rank_avg'] = df['Score'].rank(method='average', ascending=False)
df['Rank_dense'] = df['Score'].rank(method='dense', ascending=False)

# 2. Using sort_values()
df_sorted = df.sort_values(by=['Score', 'Age'], ascending=[False, True])

# 3. Using sort_index()
df_reset = df.reset_index()
df_sorted_idx = df_reset.sort_index(ascending=False)

print("Ranking:")
print(df[['Name', 'Score', 'Rank_avg', 'Rank_dense']])

print("\nSorted by Score then Age:")
print(df_sorted)

print("\nSorted by index (descending):")
print(df_sorted_idx)
```

|Method|	Description|	Duplicate Values	|Gaps in Ranks|	Start Rank|	Best Use Case|
|------|---------|-----------|----------|----------|----------|
average (default)|	Assigns average rank to tied values|	Get average of possible ranks	|Yes|	1	|Statistical analysis
min|	Uses minimum rank for tied group|	All get lowest possible rank	|Yes	|1|	Competitions (best rank)
max|	Uses maximum rank for tied group|	All get highest possible rank	|Yes	|1|	-
first|	Assigns ranks in order of appearance	|Ranked by position in data	|No	|1|	Time series data
dense|	Like 'min' but ranks increase by 1 between groups	|All get lowest possible rank	|No|	1|	Managerial reports


### **`rank()` Parameters:**

|Parameter	|Values	|Description|
|-----------|-------|-----------|

`pct=`|	`True/False`	|Return percentile ranks
`axis=`	|`0` (index), `1` (columns)	|Axis to rank along

|Parameter	|Values|	Description|

### **`sort_index()` Parameters:**
|Parameter|	Values	|Description|
|---------|---------|-----------|

### **Key Differences**
|Method	|Purpose	|Returns	|Key Parameters|	In-place Option|	Default Behavior|
|------|----------|-----------|----------|---------|-------|
`rank()`|	Assigns numerical ranks to values|	Series with same index	|`method`, `ascending`, `na_option`, `pct`	|No (returns new Series)	|`method='average'`, `ascending=True`
`sort_values()`|	Sorts data by values|	Sorted DataFrame/Series|	`by`, `ascending`, `na_position`, `key`	|`inplace=False` (default)	|`ascending=True`, `na_position='last'`
`sort_index()`|	Sorts data by index	Sorted| DataFrame/Series	|`ascending`, `na_position`, `level`	|`inplace=False` (default)	|`ascending=True`

#### Grouping

```python
# Duplicate indices are automatically grouped
obj.groupby(level=0).sum()

# Various aggregation operations can be performed
obj.groupby(level=0).mean()  # Average
obj.groupby(level=0).first() # First value 
obj.groupby(level=0).last()  # Last value 
```

Key Points
- Many pandas functions require unique indices (like `reindex`, `pivot`)
- Data selection behaves differently depending on duplicate indices
- Recommendation: Use unique indices whenever possible
- If you have duplicate indices: Always consider output type of selection operations
- To remove duplicates: Use `reset_index()` or create multi-level index

### Correlation and Covariance

    cov(X,Y) = Σ[(xi - μx)(yi - μy)] / (n-1)
    corr(X,Y) = cov(X,Y) / (σx * σy)

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([2, 4, 6, 8, 10])

# Calculate correlation
correlation = s1.corr(s2) # Output: 1.0 (perfect correlation)
covariance = s1.cov(s2) # Output: 5.0
```

```python
# Create dates
dates = pd.date_range('2020-01-01', periods=100, freq='D')
np.random.seed(42)

# Create synthetic data for 4 stocks
price_data = pd.DataFrame({
    'AAPL': 100 + np.cumsum(np.random.randn(100) * 0.5),
    'GOOG': 200 + np.cumsum(np.random.randn(100) * 0.7),
    'IBM': 150 + np.cumsum(np.random.randn(100) * 0.3),
    'MSFT': 180 + np.cumsum(np.random.randn(100) * 0.6)
}, index=dates)

# Calculate daily returns (percentage changes)
returns = price_data.pct_change()
print("Daily Returns:\n", returns.tail())
```

```python
# Full correlation matrix
correlation_matrix = returns.corr()
print("Correlation Matrix:\n", correlation_matrix)

# Important parameters:
# Correlation calculation method
corr_pearson = returns.corr(method='pearson')   # Default
corr_spearman = returns.corr(method='spearman') # Spearman correlation
corr_kendall = returns.corr(method='kendall')   # Kendall correlation

# Minimum number of non-NA observations
corr_min = returns.corr(min_periods=50)  # Minimum 50 shared data points
```

```python
# Full covariance matrix
covariance_matrix = returns.cov()
print("Covariance Matrix:\n", covariance_matrix)

# Degrees of freedom parameter
cov_ddof0 = returns.cov(ddof=0)  # Divide by n
cov_ddof1 = returns.cov(ddof=1)  # Divide by n-1 (default)
```

#### Correlation with another

```python
# Correlation of each column with a specific Series
ibm_correlations = returns.corrwith(returns['IBM'])
print("Correlations with IBM:\n", ibm_correlations)

# Correlation with another DataFrame
# Assume volume is trading volume data
volume_data = pd.DataFrame({
    'AAPL': np.random.randint(1000000, 5000000, 100),
    'GOOG': np.random.randint(500000, 3000000, 100),
    'IBM': np.random.randint(200000, 1500000, 100),
    'MSFT': np.random.randint(800000, 4000000, 100)
}, index=dates)

price_volume_corr = returns.corrwith(volume_data)
print("Correlation between returns and volume:\n", price_volume_corr)

# Row-wise correlation (axis=1)
returns_transposed = returns.T  # Transpose for example
row_correlations = returns.corrwith(returns_transposed, axis=1)
```

#### Filtering based on correlation

```python
# Finding stock pairs with high correlation
corr_matrix = returns.corr()

# Convert matrix to long format
corr_pairs = corr_matrix.stack().reset_index()
corr_pairs.columns = ['Stock1', 'Stock2', 'Correlation']

# Remove self-correlations and duplicates
corr_pairs = corr_pairs[corr_pairs['Stock1'] != corr_pairs['Stock2']]
corr_pairs = corr_pairs.drop_duplicates(subset=['Correlation'])

# Filter strong correlations
strong_corr = corr_pairs[abs(corr_pairs['Correlation']) > 0.7]
print("Strong correlations:\n", strong_corr)
```

```python
# day rolling correlation
rolling_corr = returns['AAPL'].rolling(window=20).corr(returns['MSFT'])
print("Rolling correlation of AAPL and MSFT:\n", rolling_corr.tail())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# ایجاد heatmap / Create heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, 
            annot=True, 
            cmap='coolwarm', 
            center=0,
            square=True,
            linewidths=0.5)
plt.title('Stock Correlation Matrix')
plt.show()
```

```python
# Correlation only on days with positive returns
positive_days = returns[returns['AAPL'] > 0]
positive_corr = positive_days[['AAPL', 'MSFT']].corr().iloc[0, 1]
print(f"Correlation on positive days: {positive_corr:.4f}")

# Correlation only on days with negative returns
negative_days = returns[returns['AAPL'] < 0]
negative_corr = negative_days[['AAPL', 'MSFT']].corr().iloc[0, 1]
print(f"Correlation on negative days: {negative_corr:.4f}")
```

### Unique Values, Value Counts, and Membership

```python
# Creating a sample Series
obj = pd.Series(['c', 'a', 'd', 'a', 'a', 'b', 'b', 'c', 'c'])

# Extracting unique values from Series
uniques = obj.unique()
print(uniques)

# Sorting unique values if needed
sorted_uniques = np.sort(uniques)
print(sorted_uniques)  # ['a' 'b' 'c' 'd']

# Counting frequency of each value
value_counts = obj.value_counts()
print(value_counts)
```

```python
# Using value_counts as a top-level function
print(pd.Series(obj).value_counts())

# Getting percentage instead of count
obj.value_counts(normalize=True)
```

```python
# Checking membership in a set of values
mask = obj.isin(['b', 'c'])
print(mask)

# Filtering using the boolean mask
filtered = obj[mask]
print(filtered)

# Creating sample Series for matching
to_match = pd.Series(['c', 'a', 'b', 'b', 'c', 'a'])
unique_vals = pd.Series(['c', 'b', 'a'])

# Getting index positions
indices = pd.Index(unique_vals).get_indexer(to_match)
print(indices)
```

|Method|	Description	|
|------|----------------|
`isin()`	|Compute boolean array indicating membership
`match()`|	Compute integer indices for matching
`unique()`|	Compute array of unique values
`value_counts()`|	Return Series with value frequencies

```python
# Creating a sample DataFrame
data = pd.DataFrame({'Qu1': [1, 3, 4, 3, 4],
                     'Qu2': [2, 3, 1, 2, 3],
                     'Qu3': [1, 5, 2, 4, 4]})
print(data)
# Applying value_counts to all columns
result = data.apply(pd.value_counts).fillna(0)
print(result)
```
